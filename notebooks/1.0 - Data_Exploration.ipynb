```python
import matplotlib.pyplot as plt
import math
```

```python
image_folder = dataset.dataset 

unique_samples = {}
for path, label_idx in image_folder.samples:
    class_name = image_folder.classes[label_idx]
    if class_name not in unique_samples:
        unique_samples[class_name] = path
    
    if len(unique_samples) == len(image_folder.classes):
        break

num_classes = len(unique_samples)
cols = 5
rows = math.ceil(num_classes / cols)

plt.figure(figsize=(15, 3 * rows))
sorted_classes = sorted(unique_samples.keys())

for i, class_name in enumerate(sorted_classes):
    img_path = unique_samples[class_name]
    
    img = plt.imread(img_path)
    
    plt.subplot(rows, cols, i + 1)
    plt.imshow(img)
    plt.title(class_name)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
